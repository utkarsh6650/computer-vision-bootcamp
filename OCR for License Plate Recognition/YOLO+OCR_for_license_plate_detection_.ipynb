```python
from ultralytics import YOLO
import shutil,os
```

```python
# load pretrained YOLOv8 nano model
model = YOLO("yolov8n.pt")

#Train
results = model.train(data = dataset.location+"/data.yaml",
                      # use dataset path from roboflor
                      epochs = 5,
                      imgsz = 640,
                      batch = 16,
                      workers = 2,#	Uses 2 data-loading worker processes.
                      device = 0)

# Get the exact folder this run saved to
source = os.path.join(results.save_dir, "weights", "best.pt")


os.makedirs("saved_models", exist_ok=True)
destination = "saved_models/license_plate_best.pt"

shutil.copy2(source, destination)

print("Model copied successfully!")
print("Saved at:", os.path.abspath(destination))
```
